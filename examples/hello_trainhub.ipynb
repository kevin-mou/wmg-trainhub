# Hello TrainHub

这个示例子演示训练任务的四件事：读取数据集、接收提交参数、把产物写进 `OUTPUT_DIR`、把训练指标追加到 `WMG_METRICS_FILE`。
上传本文件后，选择一个数据集（没有也可以）提交训练，然后查看日志、训练指标曲线与产物。

```python
import json
import os
import pathlib

params = json.loads(os.environ.get("WMG_PARAMS") or "{}")
dataset_dir = pathlib.Path(os.environ["WMG_DATASET_DIR"])
dataset_files = sorted(p.name for p in dataset_dir.iterdir())

print("run:", os.environ["WMG_RUN_ID"])
print("params:", params)
print("dataset files:", dataset_files)
```

```python
output_dir = pathlib.Path(os.environ["WMG_OUTPUT_DIR"])
output_dir.mkdir(parents=True, exist_ok=True)

metrics_path = pathlib.Path(os.environ["WMG_METRICS_FILE"])
metrics_path.parent.mkdir(parents=True, exist_ok=True)

epochs = int(params.get("epochs", 3))
with metrics_path.open("a", encoding="utf-8") as metrics_file:
    for epoch in range(1, epochs + 1):
        loss = 0.9 / epoch
        accuracy = min(0.99, 0.55 + 0.12 * epoch)
        print(f"epoch {epoch}/{epochs}  loss={loss:.3f}  accuracy={accuracy:.3f}")
        metrics_file.write(
            json.dumps(
                {"step": epoch, "metrics": {"loss": loss, "accuracy": accuracy}},
                ensure_ascii=False,
            )
            + "\n"
        )

result = {"status": "ok", "epochs": epochs, "datasetFiles": dataset_files}
(output_dir / "result.json").write_text(
    json.dumps(result, ensure_ascii=False, indent=2), encoding="utf-8"
)
print("artifacts:", sorted(p.name for p in output_dir.iterdir()))
```
